### 1. Write a program to accept for an integer input and print a number pattern in rows where each row contains a number repeated the same number of times as its value. E.g The following number pattern is for an integer 5. [5]


```python
#Accept an integer
number = int(input('Enter an integer:'))
for count in range(1,number+1):
    print((str(count) + ' ') * count)
```

### 2. Write a Python function that takes the following list of numbers and returns the third largest value. 
####    [13, 30, 2, 11, 15, 2, 30, 15]. 
### Ensure the function handles lists with duplicate values correctly e.g., if the list is [12, 12, 5, 6], the third largest is 5. [10]

```python
#Function declaration
def third_largest_num(mylist):
    unique_numbers = list(set(mylist))
    unique_numbers.sort()
    return unique_numbers[2]

#Initialise the list
mylist = [13, 30, 2, 11, 15, 2, 30, 15]
print(third_largest_num(mylist))
```

### 3. Write a program to create a list of 20 integers, count the occurrences of each integer, and create a dictionary to store and show the count of each element. [10]

```python
#Creating the list
int_list = [12, 13, 8, 11, 2, 1, 6, 7, 13, 12, 15, 18, 2, 1, 6, 7, 18, 17, 18, 19]
print('Original List', int_list)
count = dict()

for i in int_list:
    if i in count:
        count[i]+=1
    else:
        count[i] = 1
print('Dictionary', count)
```

### 4. Write a programm to capture the id, name, surname, age and gender of 5 people and create a dataframe. Export the dataframe to an Excel file without the index. [10]

```python
import pandas as pd
ids = []
names = []
surnames = []
ages = []
genders = []
for i in range(5):
    id = input('Enter ID:')
    name = input('Enter name:')
    surname = input('Enter surname:')
    age =int(input('Enter age:'))
    gender = input('Enter gender:')
    ids.append(id)
    names.append(name)
    surnames.append(surname)
    ages.append(age)
    genders.append(gender)
data_dict = {'ID': ids, 'Name': names, 'Surname': surnames, 'Age': ages, 'Gender': genders}

df = pd.DataFrame(data_dict)
df
df.to_excel('Downloads\people.xlsx', index = False)
```

### 6. Load the 'Student Performance' dataset (sent on the group) into a dataframe called 'sp_df'. The dataset contains empty cells in certain columns. Fill in the empty cells with the median of the column in which they appear for numeric cells and with the mode for non-numeric cells. Encode non-numeric columns using one-hot encoding. Export the resulting dataframe to an Excel file for later use. [10] 

```python
import pandas as pd
sp_df = pd.read_csv('Student Performance.csv')
#Checking for columns with empty cells
sp_df.isnull().any()
```

```python
#Understanding the column data types
sp_df.info()
```

```python
#Handling empty cells in the social_media_hours column
smh_median = sp_df['social_media_hours'].median()
sp_df['social_media_hours'] = sp_df['social_media_hours'].fillna(smh_median)
#Handling empty cells in the focus_score column
fs_median = sp_df['focus_score'].median()
sp_df['focus_score'] = sp_df['focus_score'].fillna(fs_median)
#Handling empty cells in the mental_state column
ms_mode = sp_df['mental_state'].mode() [0]
sp_df['mental_state'] = sp_df['mental_state'].fillna(ms_mode)
#Handling empty cells in the sleep_hours column
sh_median = sp_df['sleep_hours'].median()
sp_df['sleep_hours'] = sp_df['sleep_hours'].fillna(sh_median)
#Handling empty cells in the physical_activity_hours column
pah_median = sp_df['physical_activity_hours'].median()
sp_df['physical_activity_hours'] = sp_df['physical_activity_hours'].fillna(pah_median)
#Handling empty cells in the learning_style column
ls_mode = sp_df['learning_style'].mode() [0]
sp_df['learning_style'] = sp_df['learning_style'].fillna(ls_mode)
#Handling empty cells in the productivity_after_midnight column
pam_median = sp_df['productivity_after_midnight'].median()
sp_df['productivity_after_midnight'] = sp_df['productivity_after_midnight'].fillna(pam_median)
#Handling empty cells in the burnout_risk column
br_median = sp_df['burnout_risk'].median()
sp_df['burnout_risk'] = sp_df['burnout_risk'].fillna(br_median)
#Handling empty cells in the consistency_score column
cs_median = sp_df['consistency_score'].median()
sp_df['consistency_score'] = sp_df['consistency_score'].fillna(cs_median)

#Encoding non-numeric columns
encoded_sp_df = pd.get_dummies(sp_df, columns = ['gender', 'city_type', 'mental_state', 'learning_style', 'career_goal', 'performance_category'])
#Export to Excel
encoded_sp_df.to_excel('Downloads\student_performance.xlsx', index =False)
```

### 8. How average exam scores vary by where students live and by gender. [10]

```python
#Import libraries
import seaborn as sb
import matplotlib.pyplot as plt
import pandas as pd
df = pd.read_csv('Student Performance.csv')
ave_exam_score = df.groupby(['gender', 'city_type']) ['final_exam_score'].mean().reset_index()
sb.barplot(data = ave_exam_score,  x = 'city_type', y = 'final_exam_score', hue = 'gender')
plt.show()
```
